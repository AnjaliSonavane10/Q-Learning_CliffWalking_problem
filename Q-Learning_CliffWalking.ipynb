```python
import gymnasium as gym
import numpy as np
import random
```

```python
Q=np.zeros([48,4])
```

## Q- Learning

```python
alpha=0.5
epsilon=0.1
gamma=0.9
episodes=500
```

```python
#policy epsilon greedy

def epsilon_greedy(state):
    if random.random()<epsilon:
        return env.action_space.sample() #Explore
    else:
        return np.argmax(Q[state])
```

```python
for episode in range(episodes):

    env=gym.make("CliffWalking-v1")

    done=False
    state,_=env.reset()

    total_reward=0
    total_steps=0

    while not done:
        action=epsilon_greedy(state)

        next_state,reward,terminated,truncated,_=env.step(action)
        done= terminated or truncated

        #Q-learning update
        Q[state, action] += alpha * (reward + gamma * np.max(Q[next_state]) - Q[state, action])
        state=next_state

        total_reward+=reward
        total_steps+=1
    
    print("Episode:",episode,"/","500","total_reward:",total_reward,"Episode length:",total_steps)
    env.close()
```

```python
#What our agent Learned
env=gym.make("CliffWalking-v1",render_mode="human")
done=False
state,_=env.reset()

total_reward=0
total_steps=0

while not done:
    action=np.argmax(Q[state])
    next_state,reward,terminated,truncated,_=env.step(action)
    done=terminated or truncated

    state=next_state

    total_reward+=reward
    total_steps+=1

print("Total reward",total_reward,"no of steps",total_steps)
env.close()
```
