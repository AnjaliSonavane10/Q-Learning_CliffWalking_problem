## SARSA

```python
#parameters
alpha=0.5
gamma=0.9
epsilon=0.1
episodes=500
```

```python
import numpy as np
#Q table for all actions
Q=np.zeros([48,4])
```

```python
import random

#epslion gready policy

def epsilon_gready(state):
    if random.random()<epsilon:
        return env.action_space.sample() # explore
    else:
        return np.argmax(Q[state])  #exploit
```

```python
#Build sarsa

for episode in range(episodes):
    env=gym.make("CliffWalking-v1")
    
    done=False # done becomes true when process completes or truncates
    state,_=env.reset() # we get start state

    action=epsilon_gready(state)

    total_reward=0
    total_steps=0  

    while not done:
        next_state,reward,terminated,truncated,_=env.step(action) #this acually takes the action to next state
        done=terminated or truncated

        next_action=epsilon_gready(next_state)

        #update sarsa
        Q[state,action]+=alpha+(reward+gamma*(Q[next_state,next_action]-Q[state,action]))

        state=next_state
        action=next_action

        total_reward+=reward
        total_steps+=1
    
    print("Episode:",episode,"/","500","total_reward:",total_reward,"Episode length:",total_steps)
    env.close()
```

```python
#What our agent learned
env=gym.make("CliffWalking-v1",render_mode="human")
done=False
state,_=env.reset()
total_reward=0
total_steps=0

#here as agent has already learned we do not need to go acc to policy we will go with optimal Qvalue
while not done:
    action=np.argmax(Q[state])
    next_state,reward,terminated,truncated,_=env.step(action)
    done=terminated or truncated

    state=next_state

    total_reward+=reward
    total_steps+=1

print("Total reward",total_reward,"no of steps",total_steps)
env.close()
    
```
